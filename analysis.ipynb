```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = sns.load_dataset('iris')

plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='petal_length', y='petal_width', hue='species', palette='deep')
plt.title('Iris Flower Petal Length vs Width')
plt.savefig('scatter.png')
plt.close()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x='species', y='sepal_length', hue='species', palette='Set2', legend=False)
plt.title('Sepal Length Distribution by Species')
plt.savefig('chart_box.png')
plt.close()
```
